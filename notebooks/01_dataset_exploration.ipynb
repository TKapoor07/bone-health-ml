```python
from sklearn.metrics import precision_score, recall_score, f1_score

thresholds = [0.20, 0.25, 0.30, 0.35, 0.40, 0.45, 0.50]

for threshold in thresholds:
    y_pred_threshold = (y_prob >= threshold).astype(int)

    precision = precision_score(y_test, y_pred_threshold)
    recall = recall_score(y_test, y_pred_threshold)
    f1 = f1_score(y_test, y_pred_threshold)

    print(
        f"Threshold: {threshold:.2f} | "
        f"Precision: {precision:.2f} | "
        f"Recall: {recall:.2f} | "
        f"F1: {f1:.2f}"
    )
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
```

```python
df_model = femur[["SEQN", "DXXNKBMD", "T_score", "target"]].copy()

df_model = df_model.merge(
    demo[["SEQN", "RIDAGEYR", "RIAGENDR"]],
    on="SEQN",
    how="left"
)

df_model = df_model.merge(
    bmx[["SEQN", "BMXBMI"]],
    on="SEQN",
    how="left"
)

df_model = df_model.merge(
    paq[["SEQN", "PAQ650"]],
    on="SEQN",
    how="left"
)

df_model = df_model.merge(
    smq[["SEQN", "SMQ020"]],
    on="SEQN",
    how="left"
)

df_model = df_model.merge(
    alq[["SEQN", "ALQ121"]],
    on="SEQN",
    how="left"
)

df_model = df_model.merge(
    osq[["SEQN", "OSQ080", "OSQ130", "OSQ150"]],
    on="SEQN",
    how="left"
)

print(df_model.shape)
```

```python
# Convert special missing-value codes to NaN

df_model.loc[
    df_model["ALQ121"].abs() < 1e-50,
    "ALQ121"
] = float("nan")

df_model["ALQ121"] = df_model["ALQ121"].replace({
    77: float("nan"),
    99: float("nan")
})

df_model[["OSQ080", "OSQ130", "OSQ150"]] = (
    df_model[["OSQ080", "OSQ130", "OSQ150"]]
    .replace({7: float("nan"), 9: float("nan")})
)
```

```python
df_model.isnull().sum()
```

```python
df_model = df_model.dropna(subset=["target"])

print(df_model.shape)
```

```python
features = [
    "RIDAGEYR",
    "RIAGENDR",
    "BMXBMI",
    "PAQ650",
    "SMQ020",
    "ALQ121",
    "OSQ080",
    "OSQ130",
    "OSQ150"
]

X_model = df_model[features]
y_model = df_model["target"]
```

```python
print(X_model.shape)
print(y_model.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train_model, X_test_model, y_train_model, y_test_model = train_test_split(
    X_model,
    y_model,
    test_size=0.2,
    random_state=42,
    stratify=y_model
)
```

```python
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder

numeric_features = [
    "RIDAGEYR",
    "BMXBMI",
    "ALQ121"
]

categorical_features = [
    "RIAGENDR",
    "PAQ650",
    "SMQ020",
    "OSQ080",
    "OSQ130",
    "OSQ150"
]

numeric_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="median"))
])

categorical_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor_final = ColumnTransformer([
    ("numeric", numeric_transformer, numeric_features),
    ("categorical", categorical_transformer, categorical_features)
])
```

```python
from sklearn.ensemble import RandomForestClassifier

final_model = RandomForestClassifier(
    n_estimators=100,
    max_depth=5,
    min_samples_leaf=1,
    class_weight="balanced",
    random_state=42
)
```

```python
final_pipeline_clean = Pipeline([
    ("preprocessor", preprocessor_final),
    ("model", final_model)
])
```

```python
final_pipeline_clean.fit(
    X_train_model,
    y_train_model
)
```

```python
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score

y_pred_clean = final_pipeline_clean.predict(X_test_model)

y_prob_clean = final_pipeline_clean.predict_proba(
    X_test_model
)[:, 1]

print(classification_report(y_test_model, y_pred_clean))
print(confusion_matrix(y_test_model, y_pred_clean))

print(
    "ROC-AUC:",
    roc_auc_score(y_test_model, y_prob_clean)
)
```
